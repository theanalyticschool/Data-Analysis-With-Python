## Read The Data

```python
import pandas as pd
import numpy as np

import warnings
warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv(r'..\Datasets\superstore.csv')
```

```python
df.info()
```

```python
df.columns = df.columns.str.lower().str.replace(" ", "_").str.replace("-", "_")
```

```python
df.head()
```

### Region = 'West' & Sub_Category = 'Chairs' and Sales > 50

```python
df_multiple_west_all = df[(df['region'] == 'West') & (df['sub_category'] == 'Chairs') & (df['sales'] > 50)]
```

```python
df_multiple_west_all.shape
```

```python
df_multiple_west_all.head()
```

```python
df_multiple_west = df[(df['region'] == 'West') & (df['sub_category'] == 'Chairs') & (df['sales'] > 50)][['customer_id', 'region', 'sales']]
```

```python
df_multiple_west.shape
```

```python
df_multiple_west_loc = df.loc[(df['region'] == 'West') & (df['sub_category'] == 'Chairs') & (df['sales'] > 50), ['customer_id', 'region', 'sales']]
```
