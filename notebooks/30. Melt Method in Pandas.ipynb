## Read The Data

```python
import pandas as pd
import numpy as np

# Set options to display all columns and rows
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)

import warnings
warnings.filterwarnings("ignore")
```

```python
df = pd.DataFrame({'store' : ['A', 'B', 'C'],
                   'jan_sales' : [200, 220, 250],
                   'feb_sales' : [210, 230, 260],
                   'mar_sales' : [215, 240, 270]})
```

```python
df
```

## Pandas Melt Examples

```python
melt = pd.melt(df, id_vars = 'store')
```

```python
melt
```

```python
df_melt = pd.melt(df, 
                  id_vars = 'store', 
                  var_name = 'month', 
                  value_name = 'sales')
```

```python
df_melt
```

```python
df_final = pd.melt(df, id_vars = 'store', 
                   var_name = 'month', 
                   value_name = 'sales', 
                   value_vars = ['jan_sales', 'feb_sales'])
```

```python
df_final
```
