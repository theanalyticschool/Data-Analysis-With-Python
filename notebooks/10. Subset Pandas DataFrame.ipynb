### LOC Method

- Subset By Label Of The Rows & Columns
- Can Select Rows & Columns Simultaneously
<br>

### ILOC Method

- Similar To LOC Indexer But Uses Integer Location To Make The Subset

## LOC Method Syntax : df.loc[rows, cols]

**Reading A Single Column Using LOC Method**

```python
df.head()
```

```python
df.loc[: , 'segment'].head()
```

```python
df.loc[:, ['state', 'segment', 'city']].head()
```

```python
for i, j in enumerate(df.columns.tolist()):
    print(i,j)
```

```python
# df.iloc[:, ['state', 'segment', 'city']].head()
```

```python
df.iloc[: , [9, 6, 8]].head()
```
