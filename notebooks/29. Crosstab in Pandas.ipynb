## Read The Data

```python
import pandas as pd
import numpy as np

# Set options to display all columns and rows
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)

import warnings
warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv(r'..\Datasets\superstore.csv', parse_dates=['Ship Date', 'Order Date'],
                 usecols = ['Customer ID', 'Customer Name', 'Order ID', 'Order Date', 'Ship Date', 'Ship Mode',
                            'Region', 'Segment', 'Category', 'Sub-Category', 'Sales', 'Profit'])
```

```python
df.info()
```

```python
df.columns = df.columns.str.lower().str.replace(" ", "_").str.replace("-", "_")
```

```python
df.head()
```

## Crosstab Parameters

- index - vertical group column
- columns - horizontal group column

```python
df_basic = pd.crosstab(index = df['ship_mode'], columns = df['segment'])
```

```python
df_basic
```

```python
df_pivot = df.pivot_table(index = 'ship_mode', columns = 'segment', aggfunc = 'size')
```

```python
df_pivot
```

## Normalize the Frequency Table

```python
df_index = pd.crosstab(index = df['ship_mode'], columns = df['segment'], normalize = 'index').round(2) * 100
```

```python
df_index
```

```python
df_columns = pd.crosstab(index = df['ship_mode'], columns = df['segment'], normalize = 'columns').round(2) * 100
```

```python
df_columns
```

```python
df_all = pd.crosstab(index = df['ship_mode'], columns = df['segment'], normalize = 'all').round(4) * 100
```

```python
df_all
```

## Margins Parameter

```python
df_margins = pd.crosstab(index = df['ship_mode'], columns = df['segment'], margins = True)
```

```python
df_margins
```

```python
df_columns_margins = pd.crosstab(index = df['ship_mode'], columns = df['segment'], normalize = 'columns', margins = True).round(4) * 100
```

```python
df_columns_margins
```
