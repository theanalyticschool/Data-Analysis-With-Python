## Read The Data

```python
import pandas as pd
import numpy as np

import warnings
warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv(r'..\Datasets\superstore.csv', parse_dates=['Ship Date', 'Order Date'])
```

```python
df.info()
```

```python
df.columns = df.columns.str.lower().str.replace(" ", "_").str.replace("-", "_")
```

```python
df.head()
```

# Subset The Data

```python
df_subset = df.loc[:, ['customer_id', 'customer_name', 'region', 'order_date' ,'product_name', 'sales', 'quantity', 'discount']].copy()
```

```python
df_subset.head()
```

### Len Function

```python
df_subset['len_productname'] = df_subset['product_name'].str.len()
```

```python
df_subset.head()
```

### Convert Text Case

```python
df_subset['lower_region'] = df_subset['region'].str.lower()

df_subset['upper_region'] = df_subset['region'].str.upper()
```

```python
df_subset.head()
```

### Check For Substrings

```python
df_subset['contains_book'] = df_subset['product_name'].str.contains('Book')
```

```python
df_subset.head()
```

### Find & Replace

```python
df_subset['replace_region'] = df_subset['region'].str.replace("South", "SO")
```

```python
df_subset.head()
```

## Split Text

```python
df_subset['split_words'] = df_subset['product_name'].str.split()
```

```python
df_subset.head()
```

## Split Into Multiple Columns

```python
asd = df_subset['product_name'].str.split(expand = True)
```

```python
asd.head()
```

## Remove Leading & Trailing Whitespace

```python
df_subset['clean_region'] = df_subset['region'].str.strip()
```

```python
df_subset.head()
```

## Count Substrings

```python
df_subset['count_a'] = df_subset['product_name'].str.count('a')
```

```python
df_subset.head()
```

## Check Start & End

```python
df_subset['name_starts_claire'] = df_subset['customer_name'].str.startswith('Claire')

df_subset['name_ends_huff'] = df_subset['customer_name'].str.endswith('Huff')
```

```python
df_subset.head()
```

## Extract Patterns

```python
df_subset['extract_num'] = df_subset['product_name'].str.extract(r'(\d+)')
```

```python
df_subset.head()
```
