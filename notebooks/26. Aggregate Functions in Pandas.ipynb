# Importing Python Packages

```python
import pandas as pd
import numpy as np

import warnings
warnings.filterwarnings("ignore")
```

```python
data = {'product' : ['Apple', 'Banana', 'Apple', 'Banana', 'Orange', 'Orange', 'Apple', 'Cucumber'],
        'price' : [1.2, 0.5, 1.3, 0.6, 1.5, 1.6, 1.1, 2], 
        'quantity' : [75, 100, 50, 70, 60, 80, 40, 50]}

df = pd.DataFrame(data)

print(df)
```

```python
df.info()
```

# Sales Column

```python
df['sales'] = df['price'].mul(df['quantity']).astype('int')
```

```python
df.head()
```

# Calculate Total Quantity Sold, Avg Sales Price, Min & Max Selling Price

```python
total_quantity = df['quantity'].sum()
```

```python
print(total_quantity)
```

```python
avg_sales_price = df['sales'].mean()
```

```python
print(avg_sales_price)
```

```python
print(f"Min Selling Price {df['price'].min()} & Max Selling Price {df['price'].max()}")
```

# Role of Axis Parameter in Aggregate Functions
- axis = 0 : This means that the function will operate on each column and return a result for each column
- axis = 1 : This means that the function will operate on each row and return a result for each row

```python
df_col = df.drop('product', axis = 1).sum(axis = 0)
```

```python
print(df_col)
```

```python
df_row = df[['price', 'quantity']].sum(axis = 1)
```

```python
print(df_row)
```

# Applying Agg Functions on Multiple Columns

```python
avg_price_quantity = df[['price', 'quantity']].mean()
```

```python
print(avg_price_quantity)
```

# Applying Multiple Agg Functions

```python
df_agg = df[['price', 'quantity']].agg({'price' : ['sum', 'mean', 'std'],
                                        'quantity' : ['sum', 'min', 'max']})
```

```python
print(df_agg)
```

# Custom Agg Function

```python
def range_function(x): 
    
    return x.max() - x.min()
```

```python
price_range = df['price'].agg(range_function)
```

```python
print(price_range)
```
