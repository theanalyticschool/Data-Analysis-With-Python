## Read The Data

```python
import pandas as pd
import numpy as np

import warnings
warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv(r'..\Datasets\superstore.csv', parse_dates=['Ship Date', 'Order Date'],
                 usecols = ['Order ID', 'Order Date', 'Ship Date', 'Ship Mode', 'Customer ID',
                            'Customer Name', 'Segment', 'City', 'State', 'Region', 'Category',
                            'Sub-Category', 'Sales', 'Profit'])
```

```python
df.info()
```

```python
df.columns = df.columns.str.lower().str.replace(" ", "_").str.replace("-", "_")
```

```python
df.head()
```

```python
region_df = df.groupby('region')['sales'].sum()
```

```python
region_df
```

```python
region_category = df.groupby(['region', 'category'], as_index = False)['sales'].sum()
```

```python
region_category
```

```python
type(region_category)
```

```python
region_category = region_category.reset_index()

# region_category = pd.DataFrame(region_category)
```

### Multiple Aggregations

```python
region_multi = df.groupby('region').agg({'sales' : ['sum', 'min', 'max', 'mean', 'std']})
```

```python
region_multi
```

```python
region_multi.columns
```

```python
region_multi.columns = ['total_sales', 'min_sales', 'max_sales', 'mean_sales', 'std_sales']
```

```python
region_multi.reset_index(drop = False, inplace = True)
```

```python
region_multi
```

### Custom Aggregation Function

```python
def range_sales(x):
    
    return x.max() - x.min()
```

```python
region_range = df.groupby('region').agg({'sales' : ['mean', range_sales]})
```

```python
region_range.columns = ["_".join(i) if i[1] not in 'range_sales' else i[1] for i in region_range.columns]
```

```python
region_range.reset_index(drop = False, inplace = True)
```

```python
region_range
```

### Different Agg Functions on Different Columns

```python
multi_df = df.groupby('region').agg({'sales' : ['mean', 'median'], 'profit' : ['sum', 'min', 'max']})
```

```python
multi_df
```

### Named Keyword Arguements

```python
clean_df = df.groupby(['category', 'sub_category'], as_index = False).agg(totalsales = ('sales', 'sum'),
                                                                          avgsales = ('sales', 'mean'),
                                                                          medianprofit = ('profit', 'median'))
```
