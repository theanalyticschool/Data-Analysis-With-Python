**Parsing Dates For A Specific Column**

```python
df_superstore.dtypes
```

```python
df_superstore[['Order Date', 'Ship Date']] = df_superstore[['Order Date', 'Ship Date']].apply(pd.to_datetime, errors = 'coerce') 
```

```python
df_dates = pd.read_csv(r'..\Datasets\superstore.csv', parse_dates=["Order Date", "Ship Date"])
```
