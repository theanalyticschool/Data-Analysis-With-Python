## Read The Data

```python
import pandas as pd
import numpy as np

import warnings
warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv(r'..\Datasets\superstore.csv', parse_dates=['Ship Date', 'Order Date'])
```

```python
df.info()
```

```python
df.columns = df.columns.str.lower().str.replace(" ", "_").str.replace("-", "_")
```

```python
df.head()
```

# Using Assign Method

```python
df_subset = df.loc[:, ['customer_id', 'customer_name', 'region', 'order_date' ,'product_name', 'sales', 'quantity', 'discount']].copy()
```

```python
df_subset = df_subset.assign(month_order_date = lambda x : x['order_date'].dt.month,
                             year_order_date = lambda x : x['order_date'].dt.year,
                             msp = lambda x : x['sales'] / x['discount'],
                             msp_final = lambda x : np.where(x['msp'] == np.inf, x['sales'], x['msp']),
                             unit_price = lambda x : x['msp_final'] / x['quantity'], 
                             unit_price_final = lambda x : np.round(x['unit_price'],2),
                             target = lambda x : np.round(x['msp_final'] * 1.15,2)).\
                      drop(['msp', 'msp_final', 'unit_price'], axis=1).\
                      rename(columns = {'unit_price_final' : 'unit_price'})
```
