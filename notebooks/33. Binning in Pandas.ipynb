## Read The Data

```python
import pandas as pd
import numpy as np

import warnings
warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv(r'..\Datasets\superstore.csv', parse_dates=['Ship Date', 'Order Date'],
                 usecols = ['Order ID', 'Order Date', 'Ship Date', 'Ship Mode', 'Customer ID',
                            'Customer Name', 'Segment', 'City', 'State', 'Region', 'Category',
                            'Sub-Category', 'Sales', 'Profit'])
```

```python
df.info()
```

```python
df.columns = df.columns.str.lower().str.replace(" ", "_").str.replace("-", "_")
```

```python
df.head()
```

### Customers Overall Sales

```python
customers = df.groupby(['customer_id'], as_index = False).agg(totalsales = ('sales', 'sum')) \
                                                         .sort_values(by = 'totalsales', ascending = False, ignore_index = True)
```

```python
customers.head()
```

## Equal Width Binning

```python
customers['sales_bin'] = pd.cut(customers['totalsales'], bins = 10)
```

```python
customers.head()
```

```python
customers.sales_bin.value_counts()
```

## Equal Frequency Binning

```python
customers['qcut_bin'] = pd.qcut(customers['totalsales'], q=10)
```

```python
customers.head(10)
```

```python
customers.qcut_bin.value_counts()
```

## Arbitrary (Custom) Binning

```python
custom_bins = [0, 100, 500, 1000, 5000, 30000]
```

```python
labels = ['Very Low', 'Low', 'Medium', 'High', 'Very High']
```

```python
customers['custom_bins'] = pd.cut(customers['totalsales'], bins = custom_bins, labels = labels)
```

```python
customers.head()
```

```python
customers.custom_bins.value_counts()
```

## Handy Parameters

```python
cats, bins = pd.cut(customers['totalsales'], retbins = True, bins = 5)
```

```python
print(bins)
```

```python
pd.cut(customers['totalsales'], right = False, bins = 5)
```
