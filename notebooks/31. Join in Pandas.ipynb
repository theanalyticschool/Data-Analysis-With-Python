**Syntax**

```df = pd.merge(df1, df2, on = 'key_column', how = 'join_type') ```
- df1 & df2 : DataFrames
- on : The key column for joining DataFrames
- how : the type of join - inner, left, right or outer

## Read The Data

````python
import pandas as pd
import numpy as np

# Set options to display all columns and rows
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)

import warnings
warnings.filterwarnings("ignore")
````

````python
employees = pd.DataFrame({'emp_id' : [101, 102, 103, 104, 105, 106, 107],
                          'name' : ['Alice', 'Bob', 'Charlie', 'David', 'Emma', 'Frank', 'Grace'],
                          'age' : [25, 30, 28, 45, 35, 40, 29],
                          'mar_sales' : ['Durham', 'Chicago', 'Houston', 'Seattle', 'Boston', 'Denver', 'Austin']})

dept = pd.DataFrame({'emp_id' : [101, 102, 104, 108, 109],
                     'dept' : ['HR', 'Finance', 'IT', 'Sales', 'Marketing'],
                     'salary' : [60000, 75000, 72000, 50000, 65000]})
````

````python
employees
````

````python
dept
````

**Inner Join**

````python
inner_join = pd.merge(employees, dept, on = 'emp_id', how = 'inner')
````

````python
inner_join
````

**Left Join**

````python
left_join = pd.merge(employees, dept, on = 'emp_id', how = 'left')
````

````python
left_join
````

**Right Join**

````python
right_join = pd.merge(employees, dept, on = 'emp_id', how = 'right')
````

````python
right_join
````

**Outer Join**

````python
outer_join = pd.merge(employees, dept, on = 'emp_id', how = 'outer')
````

````python
outer_join
````

**Parameters** : <br>
Different Column Names - Use left_on & right_on

````python
dept.rename(columns = {'emp_id' : 'employee_id'}, inplace = True)
````

````python
dept
````

````python
new_join = pd.merge(employees, dept, how = 'inner', left_on = 'emp_id', right_on = 'employee_id')
````

````python
new_join
````

**Parameter** : indicator = True

````python
merge_join = pd.merge(employees, dept, how = 'inner', left_on = 'emp_id', right_on = 'employee_id', indicator = True)
````

````python
merge_join
````
