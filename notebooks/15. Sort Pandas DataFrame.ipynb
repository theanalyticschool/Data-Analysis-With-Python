## Read The Data

```python
import pandas as pd
import numpy as np

import warnings
warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv(r'..\Datasets\superstore.csv', parse_dates=['Ship Date', 'Order Date'])
```

```python
df.info()
```

```python
df.columns = df.columns.str.lower().str.replace(" ", "_").str.replace("-", "_")
```

```python
df.head()
```

### sort_values method

```python
df_sales_asc = df.sort_values(by = 'sales')
```

```python
df_sales_asc.head()
```

```python
df_sales_asc.tail()
```

```python
df.sales.min(), df.sales.max()
```

```python
df_sales_dsc = df.sort_values(by = 'sales', ascending = False)
```

```python
df_sales_dsc.head()
```

```python
df_sales_dsc = df.sort_values(by = 'sales', ascending = False, ignore_index = True)
```

```python
df_sales_dsc.head()
```

```python
df.head()
```

```python
df.head()
```

```python
df.sort_values(by = 'sales', ascending = False, ignore_index = True, inplace = True)
```

```python
df.head()
```

```python
df_region_sales = df.sort_values(by = ['region', 'sales'], ignore_index = True)
```

```python
df_region_sales.head()
```

```python
df_region_sales[df_region_sales.region == 'East'].head()
```

```python
df_multiple_sort = df.sort_values(by = ['region', 'sales'], ascending = [True, False], ignore_index = True)
```

```python
df_multiple_sort.head()
```

```python
df_index = df.sort_index(ascending = False)
```

```python
df_index.head()
```

```python
df_index.tail()
```

```python
df.set_index('customer_name', inplace = True)
```

```python
df.head()
```

```python
df.sort_index(ascending = True, inplace = True)
```

```python
df.head()
```

```python
df.reset_index(drop = False, inplace = True)
```

```python
df.head()
```

```python
df.sort_index(axis = 1, ascending = True, inplace = True)
```
