### Common Data Types In Pandas Dataframe:
- Integer : Whole Numbers
- Float : Decimel Numbers
- Object : Text, Mixed, Numeric & Non-Numeric Values
- Boolean : Only Two Possible Values i.e. True & False
- DateTime : Data & Time Values
- TimeDelta : Differnce Between Two DateTime Values
- Category : Finite List Of Text Values

#### Categorical Data Type

- The values should be discrete, meaning there are no partial values
- The unique values in the column should be limited & far less then the total number of values
- There shouldn't be any mysterious values that will appear in the future data of the same kind is collected

```python
df.select_dtypes('O').nunique() / df.shape[0]
```

```python
df = df.astype({'Ship Mode' : 'category', 'Segment' : 'category', 'Country' : 'category',\
                'State' : 'category', 'Region' : 'category', 'Sub-Category' : 'category',\
                'Category' : 'category'})
```

```python
df.info()
```

```python
df['Category'].str.lower().value_counts()
```

```python
df['Category'].str.capitalize().value_counts()
```
