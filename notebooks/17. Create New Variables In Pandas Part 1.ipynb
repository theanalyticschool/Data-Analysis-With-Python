## Read The Data

```python
import pandas as pd
import numpy as np

import warnings
warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv(r'..\Datasets\superstore.csv', parse_dates=['Ship Date', 'Order Date'])
```

```python
df.info()
```

```python
df.columns = df.columns.str.lower().str.replace(" ", "_").str.replace("-", "_")
```

```python
df.head()
```

# Using Square Bracket

```python
df['msp'] = df['sales'] / df['discount']
```

```python
df.head()
```

```python
df['msp'] = np.where(df['msp'] == np.inf, df['sales'], df['msp'])
```

```python
df.head()
```

# Using Assign Method

```python
df_subset = df.loc[:, ['customer_id', 'ship_mode', 'region', 'product_name', 'sales', 'quantity', 'discount', 'profit', 'msp']].copy()
```

```python
df_subset.assign(target = df['sales'] * 1.1)
```

```python
df_subset.head()
```

```python
df_subset = df_subset.assign(target = df['sales'] * 1.1)
```

```python
df_subset.head()
```

```python
df_subset = df_subset.assign(profit_perc = lambda x : x.profit / x.msp)
```

```python
df_subset.head()
```

# Using Insert Method

```python
df_subset.insert(5, 'unit_sales_price', df['sales']/df['quantity'])
```

```python
df_subset.head()
```

```python
df_subset.columns.get_loc('msp')
```

```python
df_subset.insert(df_subset.columns.get_loc('msp') + 1, 'unit_price', df['msp'] / df['quantity'])
```
