- & - and

## Read The Data

```python
import pandas as pd
import numpy as np

import warnings
warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv(r'..\Datasets\superstore.csv')
```

```python
df.info()
```

```python
df.columns = df.columns.str.lower().str.replace(" ", "_").str.replace("-", "_")
```

```python
df.head()
```

### Region = 'West' & Sub_Category = 'Chairs' and Sales > 50

```python
filter_west = df['region'] == 'West'
filter_sub = df['sub_category'] == 'Chairs'
filter_sales = df['sales'] > 50

df_multiple_west = df[filter_west & filter_sub & filter_sales]
```

```python
df_multiple_west.shape
```

```python
df_multiple_west.sub_category.unique()
```

```python
df_multiple_west.region.unique()
```

```python
df_multiple_west.sales.min()
```

```python
df_multiple_west.head()
```

```python
df_multiple_west_all = df[(df['region'] == 'West') & (df['sub_category'] == 'Chairs') & (df['sales'] > 50)]
```

```python
df_multiple_west_all.shape
```

```python
df['ship_date'] = pd.to_datetime(df['ship_date'])
```

```python
df_category = df[((df['ship_date']>='2020-01-01') & (df['ship_date']<='2020-03-31')) & (df['region'] == 'West') & (df['category'] == 'Furniture')]
```

```python
df_category.shape
```

```python
df_category.ship_date.min(), df_category.ship_date.max(), df_category.region.unique(), df_category.category.unique()
```

```python
df_not_category = df[~(((df['ship_date']>='2020-01-01') & (df['ship_date']<='2020-03-31')) & (df['region'] == 'West') & (df['category'] == 'Furniture'))]
```
