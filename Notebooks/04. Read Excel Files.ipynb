**Reading Specific Columns Using Read_Excel Method**

```python
df_sel = pd.read_excel('..\Datasets\Superstore.xls', usecols=['Customer ID', 'Sales', 'Profit'])
```

```python
df_sel.head()
```

**Specify Data Types In Read_Excel**

```python
df_data_type = pd.read_excel('..\Datasets\Superstore.xls', dtype = {'Customer ID' : 'object', 'Sales' : 'float'})
```

```python
df_data_type.head()
```

```python
df_data_type.info()
```
