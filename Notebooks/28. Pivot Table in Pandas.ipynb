## Read The Data

```python
import pandas as pd
import numpy as np

import warnings
warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv(r'..\Datasets\superstore.csv', parse_dates=['Ship Date', 'Order Date'],
                 usecols = ['Customer ID', 'Customer Name', 'Order ID', 'Order Date', 'Ship Date', 'Ship Mode',
                            'Region', 'Segment', 'Category', 'Sub-Category', 'Sales', 'Profit'])
```

```python
df.info()
```

```python
df.columns = df.columns.str.lower().str.replace(" ", "_").str.replace("-", "_")
```

```python
df.head()
```

## Pivot Table Parameters

- index - vertical group column
- columns - horizontal group column
- values - aggregating column
- aggfunc - aggregation function (deafult to the mean)

```python
df_basic = df.pivot_table(index = 'region', columns = 'sub_category', values = 'profit', aggfunc = 'sum')
```

```python
df_basic
```

## Comparison to Group By

```python
df_group = df.groupby(['region', 'sub_category']).agg(sum_profit = ('profit', 'sum'))
```

```python
df_group.head(20)
```

## Different Aggregate Functions

```python
df_default = df.pivot_table(index = 'region', columns = 'sub_category', values = 'profit', aggfunc = 'mean').round(2)
```

```python
df_default
```

```python
df_median = df.pivot_table(index = 'region', columns = 'sub_category', values = 'profit', aggfunc = 'median').round(2)
```

```python
df_median
```

## Styling Pivot Tables

```python
df_reg_cat = df.pivot_table(index = 'region', columns = 'category', values = 'profit', aggfunc = 'mean').round(2)
```

```python
df_reg_cat
```

```python
df_reg_cat.style.highlight_max()
```

```python
df_reg_cat.style.highlight_max(axis = 1)
```

```python
df_reg_cat.style.background_gradient(cmap = 'magma')
```

## Size of Each Group

```python
df_size = df.pivot_table(index = 'region', columns = 'category', aggfunc = 'size')
```

```python
df_size
```

## Row & Columns Totals

```python
df_total = df.pivot_table(index = 'region', columns = 'category', values = 'sales' , aggfunc = 'mean', margins = True).round(2)
```

```python
df_total
```

```python
df['sales'].mean()
```

## More Than Two Grouping Columns

```python
df_more_pivot = df.pivot_table(index = ['region', 'segment'], columns = 'category', values = 'sales', aggfunc = 'max').round(2)
```

```python
df_more_pivot
```

```python
df_multi_col = df.pivot_table(index = ['region'], columns = ['category','sub_category'], values = 'sales', aggfunc = 'max').round(2)
```

```python
df_multi_col
```

## Multiple Aggregations

```python
df_agg = df.pivot_table(index = ['region'], columns = ['segment'], values = 'sales', aggfunc = ['max', 'min', 'mean']).round(2)
```

```python
df_agg
```
