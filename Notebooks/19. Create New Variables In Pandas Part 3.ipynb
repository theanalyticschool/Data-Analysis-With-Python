## Read The Data

```python
import pandas as pd
import numpy as np

import warnings
warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv(r'..\Datasets\superstore.csv', parse_dates=['Ship Date', 'Order Date'])
```

```python
df.info()
```

```python
df.columns = df.columns.str.lower().str.replace(" ", "_").str.replace("-", "_")
```

```python
df.head()
```

## Subset The Data

```python
df_subset = df.loc[:, ['customer_id', 'customer_name', 'region', 'order_date' ,'product_name', 'sales', 'quantity', 'discount']].copy()
```

## np.where()

```python
df_subset['sales_1000'] = np.where(df_subset['sales'] >= 1000, "Yes", "No")
```

```python
df_subset.head()
```

```python
df_subset.query('sales_1000 == "Yes"')['customer_id'].nunique()
```

```python
df_subset['customer_id'].nunique()
```

## np.select()

```python
conditions = [df_subset['sales'] < 100, df_subset['sales'].between(100, 300), df_subset['sales'].between(300, 1000), df_subset['sales'] > 1000]

choices = ['low', 'average', 'above_average', 'high']

df_subset['sales_bin'] = np.select(conditions, choices, default = 'NA')
```
