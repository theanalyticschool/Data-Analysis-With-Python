## Read The Data

````python
import pandas as pd
import numpy as np

# Set options to display all columns and rows
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)

import warnings
warnings.filterwarnings("ignore")
````

## Example 1 : Combining Monthly Sales (Vertical Concat)

````python
# Monthly sales data for three months
sales_jan = pd.DataFrame({
                            'Month': ['Jan'] * 3,
                            'Region': ['East', 'West', 'North'],
                            'Revenue': [12000, 15000, 13000]
                        })

sales_feb = pd.DataFrame({
                            'Month': ['Feb'] * 3,
                            'Region': ['East', 'West', 'North'],
                            'Revenue': [14000, 16000, 12500]
                        })

sales_mar = pd.DataFrame({
                            'Month': ['Mar'] * 3,
                            'Region': ['East', 'West', 'North'],
                            'Revenue': [13500, 15500, 14500]
                        })
````

````python
sales_q1 = pd.concat([sales_jan, sales_feb, sales_mar], ignore_index = True)
````

````python
sales_q1
````

## Example 2 : Adding More Details (Horizontal Concat)

````python
# Customer data
customers = pd.DataFrame({
                            'CustomerID': [101, 102, 103],
                            'CustomerName': ['Alice', 'Bob', 'Charlie']
                        })

# Corresponding region data
regions = pd.DataFrame({
                        'Region': ['East', 'West', 'North']
                      })
````

````python
customer_details = pd.concat([customers, regions], axis = 1)
````

````python
customer_details
````

## Example 3 : Combining Online & Store Sales with Different Columns

````python
# Online sales data
online_sales = pd.DataFrame({
                             'OrderID': [1, 2, 3],
                             'Channel': ['Online', 'Online', 'Online'],
                             'Revenue': [500, 700, 650]
                           })

# Store sales data
store_sales = pd.DataFrame({
                            'OrderID': [4, 5],
                            'Channel': ['Store', 'Store'],
                            'Discount': [50, 30],
                            'Revenue': [800, 600]
                          })
````

````python
all_sales = pd.concat([online_sales, store_sales], axis = 0, ignore_index = True)
````

````python
all_sales
````

## Example 4 : Labeling Data with Keys

````python
sales_with_keys = pd.concat([sales_jan, sales_feb, sales_mar], keys = ['Jan', 'Feb', 'Mar'])
````
