## Read The Data

```python
import pandas as pd
import numpy as np

import warnings
warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv(r'..\Datasets\superstore.csv', parse_dates=['Ship Date', 'Order Date'])
```

```python
df.info()
```

```python
df.columns = df.columns.str.lower().str.replace(" ", "_").str.replace("-", "_")
```

```python
df.head()
```

## Subset The Data

```python
df_subset = df.loc[:, ['customer_id', 'customer_name', 'region', 'order_date' ,'product_name', 'sales', 'quantity', 'discount']].copy()
```

```python
df_subset['email_sales'] = df_subset['sales'].apply(lambda x : 'Yes' if x > 100 else 'No')
```

```python
df_subset.head()
```

```python
df_subset = pd.concat((df_subset, df['profit']), axis = 1)
```

```python
df_subset.head()
```

```python
def customer_type(x):
    
    if x <=0:
        
        return 'low'
    
    elif x>0 and x <= 10:
        
        return 'mid'
    
    else:
        
        return 'high'
```

```python
df_subset['customer_type'] = df_subset['profit'].apply(customer_type)
```

```python
df_subset.head()
```

```python
df_subset.customer_type.value_counts()
```

```python
def unit_price(x):
    
    return ((x['sales'] / x['quantity']) / (1 - x['discount']))
```

```python
df_subset['unitprice'] = df_subset.apply(unit_price, axis = 1)
```

```python
df_subset.head()
```
