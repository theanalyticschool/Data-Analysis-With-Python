### Common Data Types In Pandas Dataframe:
- Integer : Whole Numbers
- Float : Decimel Numbers
- Object : Text, Mixed, Numeric & Non-Numeric Values
- Boolean : Only Two Possible Values i.e. True & False
- DateTime : Data & Time Values
- TimeDelta : Differnce Between Two DateTime Values
- Category : Finite List Of Text Values

### Read Data

```python
import pandas as pd
import numpy as np

import warnings
warnings.filterwarnings("ignore")
```

```python
df = pd.DataFrame({
                   'customerid' : ['1', 2, '3', '4', 5],
                   'quantity' : [1, 2.0, 3.7, 4.99, 5],
                   'region' : ['east', 'west', 3, 4, np.nan],
                   'gender' : ['M', 'M', 'F', 'M', 'F'],
                   'discount' : ['21%', '10%', '15%', '18%', '20%'],
                   'sales' : [10.1, 11.2, 21.3, '24.5', 'north'],
                   'profit' : ['$2', '$4.1', '$5.6', '$1.7', '$8.1'],
                   'ct' : [1, 0, 1, 0, 1]})
```

```python
df.dtypes
```

```python
df.info()
```

## AS TYPE METHOD

```python
df
```

```python
df = df.astype({'customerid' : 'O', 'quantity' : 'int'})
```

```python
df.info()
```

```python
df
```

```python
df['ct'] = df['ct'].astype('bool')

df['gender_wrong'] = df['gender'].astype('bool')
```

```python
df.info()
```

```python
df
```

```python
df['gender_correct'] = np.where(df['gender'] == 'F', 1, 0)

df['gender_correct'] = df['gender_correct'].astype('bool')
```

```python
df.info()
```

```python
df
```

```python
df['cusid'] = df['customerid'].astype('float')
```

```python
df.info()
```

```python
df
```

```python
df['profit'] = df['profit'].apply(lambda x : x.replace("$", "")).astype('float')
```

```python
df.info()
```

```python
df
```

```python
df['discount'] = df['discount'].apply(lambda x : x.replace("%", "")).astype('float')
```

```python
df['discount'] = df['discount'].div(100, 1)
```

```python
df
```

## TO_NUMERIC

```python
df['sales_one'] = df['sales'].replace(r"[a-z]+", np.nan, regex = True).astype('float')
```

```python
df
```

```python
df['sales_two'] = pd.to_numeric(df['sales'], errors = 'coerce')
```

```python
df.info()
```
