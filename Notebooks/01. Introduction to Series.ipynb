## Creating Series Using List

```python
df_list = pd.Series(data = [10, 20, 30, 40, 50])
```

```python
df_list
```

```python
df_list_custom = pd.Series(data = [10, 20, 30, 40], index = ['SA', 'SB', 'SC', 'SD'])
```

```python
df_list_custom
```

## Create A Series Using Dict

```python
df_dict = pd.Series(data = {'SA' : 10, 'SB' : 20, 'SC' : 30, 'SD' : 40})
```

```python
df_dict
```

```python
df_dict.equals(df_list_custom)
```

```python
df_dict = pd.concat([df_dict, pd.Series({'SD': 40})])
```

```python
df_dict
```

```python
df_dict_append = pd.concat([df_dict, df_list_custom])
```

```python
df_dict_append
```
