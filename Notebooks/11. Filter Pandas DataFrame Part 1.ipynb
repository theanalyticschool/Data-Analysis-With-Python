# Rules For []
- string - return a column as a series
- list of strings - return a dataframe with all those columns
- sequence of booleans - return all rows that are True

```python
df['ship_date'] = pd.to_datetime(df['ship_date'])
```

```python
filter_shipdate = df['ship_date'] >= '2021-01-01'

df_filter_ship = df[filter_shipdate]
```
