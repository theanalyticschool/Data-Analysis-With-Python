```python
import pandas as pd
```

```python
df = pd.read_csv(r'..\Datasets\orders.csv', parse_dates=['Order Date'], usecols=['Customer ID', 'Order Date', 'Order ID'])

df.columns = df.columns.str.lower().str.replace(' ', '_')

df = df.loc[:, ['customer_id', 'order_date', 'order_id']]

df.sort_values(by = ['customer_id', 'order_date', 'order_id'], ignore_index=True, inplace=True)
```

```python
df.head(10)
```

### **Check For Exact Duplicate Rows**

```python
duplicate_count = df.duplicated().sum()
```

```python
print(duplicate_count)
```

### **View Duplicate Rows**

```python
duplicates = df[df.duplicated()]
```

```python
duplicates.head(10)
```

### **Check Duplicates Based on Duplicate Columns**

```python
order_duplicates = df.duplicated(subset=['order_id']).sum()
```

```python
print(order_duplicates)
```

### **Inspect Duplicate Groups**

```python
duplicate_orders = df[df.duplicated(subset = ['order_id'], keep=False)].sort_values(by='order_id')
```

```python
duplicate_orders.head(20)
```

### **Remove Exact Duplicate Rows**

```python
df_dedup_exact = df.drop_duplicates()
```

```python
print(df.shape)
```

```python
print(df_dedup_exact.shape)
```

### **Keep Last Occurence**

```python
df_order_last = df.drop_duplicates(subset = ['order_id'], keep='last')
```

```python
print(df_order_last.shape)
```

### **Remove All Duplicate Groups**

```python
df_unique_orders = df.drop_duplicates(subset = 'order_id', keep = False)
```

```python
print(df_unique_orders.shape)
```

### **Flag Duplicates**

```python
df['is_duplicate_order_id'] = df.duplicated(subset = 'order_id', keep =False)
```
