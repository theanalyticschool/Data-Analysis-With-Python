## Read The Data

```python
import pandas as pd
import numpy as np

import warnings
warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv(r'..\Datasets\superstore.csv', parse_dates=['Ship Date', 'Order Date'])
```

```python
df.info()
```

```python
df.columns = df.columns.str.lower().str.replace(" ", "_").str.replace("-", "_")
```

```python
df.head()
```

# Subset The Data

```python
df_subset = df.loc[:, ['customer_id', 'customer_name', 'region', 'order_date' ,'product_name', 'sales', 'quantity', 'discount']].copy()
```

```python
df_numerical = df.select_dtypes(include = 'number')
```

```python
df_numerical.head()
```

```python
df_numerical.drop('postal_code', axis = 1, inplace = True)
```

```python
df_numerical.head()
```

```python
df_num_square = df_numerical.map(lambda x : x**2)
```

```python
df_num_square.head()
```

```python
df_subset.head()
```

```python
def custom_capitalize(x):
    
    if isinstance(x, str):
        
        return x.upper()
    
    else:
        
        return x
```

```python
df_subset = df_subset.map(custom_capitalize)
```
