## Read The Data

```python
import pandas as pd
import numpy as np

import warnings
warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv(r'..\Datasets\sales.csv')
```

```python
df.info()
```

```python
df
```

### <font color = 'grey'>Addition</font>

```python
df['SALES_2018_2019'] = df['SALES_2018'] + df['SALES_2019']
```

```python
df.head()
```

```python
df['SALES_2021_2022'] = df['SALES_2021'] + df['SALES_2022']
```

```python
df.head()
```

```python
df['NEW_SALES_2021_2022'] = df['SALES_2021'].add(df['SALES_2022'], fill_value = 0)
```

```python
df.head()
```

### <font color = 'grey'>Multiplication</font>

```python
df['mul_2020'] = df['SALES_2020'] * 1.2
```

```python
df.head()
```

```python
df['mul_2020_new'] = df['SALES_2020'].mul(10, fill_value = 10)
```

```python
df.head()
```

### <font color = 'grey'>Subtraction</font>

```python
df['var_2021_2020'] = df['SALES_2021'].sub(df['SALES_2020'])
```

```python
df.head()
```

### <font color = 'grey'>Division</font>

```python
df['div_sales_10'] = df['SALES_2021'].div(10, fill_value = 1)
```

```python
df.head()
```

```python
df['SALES_2018_ABS'] = df['SALES_2018'].abs()
```

```python
df.head()
```

```python
df_new = df.round(2)
```

```python
df_new.head()
```

```python
df1 = df.round({'SALES_2019' : 1, 'SALES_2020' : 2})
```
