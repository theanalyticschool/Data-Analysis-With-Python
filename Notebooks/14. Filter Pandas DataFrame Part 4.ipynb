## Read The Data

```python
import pandas as pd
import numpy as np

import warnings
warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv(r'..\Datasets\superstore.csv')
```

```python
df.info()
```

```python
df.columns = df.columns.str.lower().str.replace(" ", "_").str.replace("-", "_")
```

```python
df.head()
```

### Filter Superstore Dataset for sales > 100 - Boolean Mask Indexing

```python
df_boolean = df[df['sales']>100].reset_index(drop = True).copy()
```

```python
df.shape, df_boolean.shape
```

### Filter Superstore Dataset for sales > 100 - Query Method

```python
df.query('sales > 100').head()
```

```python
df.query('100 <= sales <= 200').head()
```

### Filter Superstore Dataset Where Segment Is Conusmer & Sales >100

```python
df.query('(segment == "Consumer") and (sales > 100)').head()
```

```python
df.query('(segment == "Consumer") and (sales > 100)').head()
```

```python
df.query('profit > sales').head()
```

### Use Of IN Operator In Query Method

```python
df.query('region in ["East", "West"]').head()
```

```python
df.query('region not in ["East", "West"]').head()
```

```python
df.query('sales + profit >500').head()
```

```python
min_sales = 750

df.query('sales > @min_sales').head()
```

```python
df.query('index >9000 and sales>500').head()
```
