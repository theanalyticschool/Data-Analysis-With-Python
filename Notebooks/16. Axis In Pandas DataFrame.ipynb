## Read The Data

```python
import pandas as pd
import numpy as np

import warnings
warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv(r'..\Datasets\superstore.csv', parse_dates=['Ship Date', 'Order Date'])
```

```python
df.info()
```

```python
df.columns = df.columns.str.lower().str.replace(" ", "_").str.replace("-", "_")
```

```python
df.head()
```

```python
df_pivot = pd.pivot_table(data = df, index = 'sub_category', columns = 'region', values = 'sales', aggfunc = 'sum').round(0).astype('int')

df_pivot.columns.name = ""
```

```python
df_pivot.shape
```

## Axis = 0
Action Will Happen On Rows For Each And Every Column
<br>------------<br>
</n>Objective : To Find The Sum Of Sales For Each Region

```python
df_pivot.head()
```

```python
df_pivot.sum(axis = 0)
```

## Axis = 1
- Action Will Happen On Columns For Each And Every Row
<br>------------<br>
</n>Objective : To Find The Sum Of Sales For Each Sub-Category

```python
df_pivot.sum(axis = 1)
```
