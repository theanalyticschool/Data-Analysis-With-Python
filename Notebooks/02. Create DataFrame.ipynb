```python
import pandas as pd
```

**Dictionaries**

```python
revenue = [{'company' : 'apple', 'jan' : 20, 'feb' : 30, 'mar' : 40}, 
           {'company' : 'amazon', 'jan' : 10, 'feb' : 30, 'mar' : 50}, 
           {'company' : 'microsoft', 'jan' : 30, 'feb' : 25, 'mar' : 20}]
```

```python
df_list = pd.DataFrame(revenue)
```

```python
df_list.head()
```

```python
revenue2 = {'company' : ['apple', 'amazon', 'microsoft'], 
            'jan' : [20, 10, 30], 
            'feb' : [30, 30, 25], 
            'mar' : [40, 50, 20]}
```

```python
df_rev2 = pd.DataFrame(revenue2)
```

```python
df_rev2.head()
```

**Lists**

```python
revenue3 = [('apple', 20, 30, 40), 
            ('amazon', 10, 30, 50), 
            ('microsoft', 30, 25, 20)]

labels = ['company', 'jan', 'feb', 'mar']
```

```python
df = pd.DataFrame(revenue3, columns = labels)
```
