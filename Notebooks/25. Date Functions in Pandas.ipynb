# Importing Python Packages

```python
import pandas as pd
import numpy as np

import warnings
warnings.filterwarnings("ignore")
```

```python
data = {'Date' : ['2024-01-01', '2024-02-15', '2024-03-10', '2024-04-25'],
        'Value' : [100, 200, 300, 400]}

df = pd.DataFrame(data)

print(df)
```

```python
df.info()
```

# Convert String Dates to Datetime Data Type in Pandas

```python
df['Date'] = pd.to_datetime(df['Date'])
```

```python
df.info()
```

```python
print(df)
```

# Extracting Date Components Using .dt Accessor

```python
df['Year'] = df['Date'].dt.year

df['Month'] = df['Date'].dt.month

df['Day'] = df['Date'].dt.day
```

```python
print(df)
```

# Calculate Date Differences

```python
firstdate = df['Date'].min()
```

```python
print(firstdate)
```

```python
df['Days_Since_Start'] = (df['Date'] - firstdate).dt.days
```

```python
print(df)
```

# Generate Date Ranges

```python
date_range = pd.date_range(start = '2024-01-01', end = '2024-12-31', freq = 'ME')
```

```python
print(date_range)
```

# Extracting Monthly Information

```python
df_sales = pd.DataFrame({'Date' : pd.date_range(start = '2024-01-01', periods = 100, freq = 'D'), 
                        'Sales' : range(100)})
```

```python
print(df_sales.head())
```

```python
df_sales['monthly_period'] = df_sales['Date'].dt.to_period('M').dt.to_timestamp()
```

```python
print(df_sales.tail())
```

# Handling Time Zones

```python
df_sales['Date'] = df_sales['Date'].dt.tz_localize('UTC')
```

```python
df_sales.head()
```

```python
df_sales['NYDate'] = df_sales['Date'].dt.tz_convert('America/New_York')
```
