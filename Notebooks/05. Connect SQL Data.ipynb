## Additional Details
- **PyODBC Python Package**
- **Database Installed such as SQL Server**
- **Datasets in Database**

````python
import pandas as pd
import numpy as np
import pyodbc
import warnings

warnings.filterwarnings("ignore")
````

````python
pyodbc.drivers()
````

## Trusted Connection

````python
trusted_conn = pyodbc.connect('DRIVER={ODBC Driver 17 for SQL Server};'
                              'SERVER=XXXXX;'
                              'DATABASE=XXXXX;'
                              'Trusted_Connection=yes;')
````

## Non-Trusted Connection

```python
non_trusted_conn = pyodbc.connect('DRIVER={ODBC Driver 17 for SQL Server};'
                                  'SERVER=XXXXX;'
                                  'DATABASE=XXXXX;'
                                  'UID=XXXX;'
                                  'PWD=XXXX;')
```

**Read Data From SQL Server**

````python
df_orders = pd.read_sql('Select * from dbo.orders', trusted_conn)
````

````python
df_returns = pd.read_sql('Select * from dbo.returns', trusted_conn)
````
