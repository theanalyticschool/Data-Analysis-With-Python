## Read The Data

```python
import pandas as pd
import numpy as np

import warnings
warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv(r'..\Datasets\superstore.csv', parse_dates=['Ship Date', 'Order Date'])
```

```python
df.info()
```

```python
df.columns = df.columns.str.lower().str.replace(" ", "_").str.replace("-", "_")
```

```python
df.head()
```

# Subset The Data

```python
df_subset = df.loc[:, ['customer_id', 'customer_name', 'region', 'order_date' ,'product_name', 'sales', 'quantity', 'discount']].copy()
```

```python
df_subset['square_sales'] = df_subset['sales'].map(lambda x : x **2)
```

```python
df_subset.head()
```

```python
mapping = {'Bookcases' : 'Furniture', 'Chairs' : 'Furniture', 'Furnishings' : 'Furniture', 'Tables' : 'Furniture',
           'Appliances' : 'Office', 'Art' : 'Office', 'Binders' : 'Office', 'Envelopes' : 'Office', 'Fasteners' : 'Office', 
           'Labels' : 'Office', 'Paper' : 'Office', 'Storage' : 'Office', 'Supplies' : 'Office', 'Accessories' : 'Technology', 
           'Copiers' : 'Technology', 'Machines' : 'Technology', 'Phones' : 'Technology'}
```

```python
df['new_category'] = df['sub_category'].map(mapping)
```
